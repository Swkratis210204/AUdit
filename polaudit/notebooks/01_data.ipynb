# 1. Load and clean a dataset

The toy dataset has 60 texts rated by 24 annotators. Twenty texts are consensus, twenty are polarized by politics, twenty by gender.

```python
import tempfile
from pathlib import Path

import pandas as pd

from polaudit import run_dataset
from polaudit.toy import toy_ratings, toy_spec

pd.set_option("display.width", 160)
result = run_dataset(toy_spec(), raw=toy_ratings())        # the toy dataset: 60 texts, 24 annotators, gender / politics / age
```

```python
print(result.describe())
result.groups
```

## Your own dataset

A `DatasetSpec` names the columns and maps raw values to groups. A value you do not list is removed.

```python
from polaudit import DatasetSpec

raw = toy_ratings()
my_spec = DatasetSpec(
    name="mydata",
    load=lambda root=None: raw,
    dimensions={"politics": ("politics", {"left": "left", "right": "right"})},      # only politics, and only these two values
    min_au_texts=5,
)
mine = run_dataset(my_spec, raw=raw)
mine.per_dimension
```
